# Convert dates

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

```python
df[["Order Date", "Ship Date"]].head()
```

# Create useful business features

```python
df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month
df["Month_Name"] = df["Order Date"].dt.month_name()
df["Quarter"] = df["Order Date"].dt.quarter
```

```python
df["Shipping_Days"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days
```

```python
df[[
    "Order Date",
    "Ship Date",
    "Shipping_Days"
]].head()
```

# Sales and profit overview

```python
print("Total Sales:", df["Sales"].sum())
print("Total Profit:", df["Profit"].sum())
print("Total Quantity:", df["Quantity"].sum())
print("Total Orders:", df["Order ID"].nunique())
print("Total Customers:", df["Customer ID"].nunique())
```

# Calculate profit margin

```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()

profit_margin = (total_profit / total_sales) * 100

print(f"Profit Margin: {profit_margin:.2f}%")
```

```python
import pandas as pd

df_mysql = pd.read_csv(
    r"D:\superstore.csv",
    encoding="latin1"
)

print("Rows:", len(df_mysql))
print("Columns:", len(df_mysql.columns))
```

```python
print(df_mysql.columns.tolist())
```

```python
import mysql.connector

conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="NewPassword@123",
    database="superstore_db"
)

cursor = conn.cursor()

print("Connected to MySQL successfully!")
```

```python
df_mysql["Order Date"] = pd.to_datetime(
    df_mysql["Order Date"],
    dayfirst=True
)

df_mysql["Ship Date"] = pd.to_datetime(
    df_mysql["Ship Date"],
    dayfirst=True
)

print(df_mysql[["Order Date", "Ship Date"]].head())
```

```python
import math

insert_query = """
INSERT INTO superstore_raw (
    `Row ID`, `Order ID`, `Order Date`, `Ship Date`,
    `Ship Mode`, `Customer ID`, `Customer Name`, `Segment`,
    `Country/Region`, `City`, `State/Province`, `Postal Code`,
    `Region`, `Product ID`, `Category`, `Sub-Category`,
    `Product Name`, `Sales`, `Quantity`, `Discount`, `Profit`
)
VALUES (
    %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s,
    %s, %s, %s, %s, %s, %s, %s, %s, %s, %s
)
"""

data = []

for row in df_mysql.itertuples(index=False, name=None):
    cleaned_row = tuple(
        None if (isinstance(value, float) and math.isnan(value)) else value
        for value in row
    )
    data.append(cleaned_row)

cursor.executemany(insert_query, data)
conn.commit()

print("Rows inserted:", cursor.rowcount)
```

```python
# Convert dates to Python date objects
df_mysql["Order Date"] = pd.to_datetime(df_mysql["Order Date"]).dt.date
df_mysql["Ship Date"] = pd.to_datetime(df_mysql["Ship Date"]).dt.date

print(df_mysql[["Order Date", "Ship Date"]].head())
```

```python
insert_query = """
INSERT INTO superstore (
    `Row ID`, `Order ID`, `Order Date`, `Ship Date`,
    `Ship Mode`, `Customer ID`, `Customer Name`, `Segment`,
    `Country/Region`, `City`, `State/Province`, `Postal Code`,
    `Region`, `Product ID`, `Category`, `Sub-Category`,
    `Product Name`, `Sales`, `Quantity`, `Discount`, `Profit`
)
VALUES (
    %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s,
    %s, %s, %s, %s, %s, %s, %s, %s, %s, %s
)
"""

data = []

for row in df_mysql.itertuples(index=False, name=None):
    data.append(row)

cursor.executemany(insert_query, data)
conn.commit()

print("Rows inserted:", cursor.rowcount)
```
